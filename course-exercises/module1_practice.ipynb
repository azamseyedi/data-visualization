```python
import matplotlib.pyplot as plt
import numpy as np

# ساخت داده‌ها
data = np.random.normal(loc=50, scale=10, size=1000)

# رسم هیستوگرام
plt.hist(data, bins=30, color='skyblue', edgecolor='black')
plt.title('Normal Distribution (Mean=50, Std=10)')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()

print("Mean:", np.mean(data))
print("Std Dev:", np.std(data))
print("Min:", np.min(data))
print("Max:", np.max(data))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# رسم هیستوگرام با Seaborn (همراه با منحنی تخمین چگالی یا kde)
sns.histplot(data, bins=30, kde=True, color="purple")

plt.title('Normal Distribution (Mean=50, Std=10)')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()

print("Mean:", np.mean(data))
print("Std Dev:", np.std(data))
print("Min:", np.min(data))
print("Max:", np.max(data))
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# ساخت ۱۰۰۰ داده با توزیع پواسون و نرخ وقوع 5
poisson_data = np.random.poisson(lam=5, size=1000)

# رسم هیستوگرام بدون منحنی پیوسته (kde=False)
sns.histplot(poisson_data, kde=False, discrete=True, color="mediumseagreen")
plt.title("Poisson Distribution (lam=5)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# انجام آزمایش (۱۰ بار پرتاب سکه) به تعداد ۱۰۰۰ بار
binomial_data = np.random.binomial(n=10, p=0.5, size=1000)

# رسم هیستوگرام گسسته با Seaborn
sns.histplot(binomial_data, discrete=True, color="royalblue")
plt.title("Binomial Distribution (n=10, p=0.5, size=1000)")
plt.xlabel("Number of Successes (Heads)")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# ساخت ۱۰۰۰ داده با توزیع برنولی (n=1)
bernoulli_data = np.random.binomial(n=1, p=0.5, size=1000)

# رسم هیستوگرام با Seaborn
sns.histplot(bernoulli_data, discrete=True, color="crimson")
plt.title("Bernoulli Distribution (p=0.5, size=1000)")
plt.xlabel("Outcome (0 = Failure, 1 = Success)")
plt.ylabel("Frequency")
plt.xticks([0, 1])  # تنظیم محور افقی فقط روی ۰ و ۱
plt.show()
```

```python
import matplotlib.pyplot as plt

scores = [40, 45, 50, 52, 55, 60, 65, 70, 80, 95]

fig, ax = plt.subplots()

# تنظیم Box Plot به صورت افقی با whisker روی صدک‌های ۲۰ و ۸۰
ax.boxplot(scores, vert=False, whis=(20, 80), sym="go")

ax.set_title("Horizontal Box Plot (20th - 80th Percentile)")
ax.set_xlabel("Scores")

plt.show()
```

```python
from sklearn.datasets import make_blobs

X, y = make_blobs(n_samples=500, centers=2, random_state=42)

print(X.shape)
print(y.shape)

blob_0 = X[y == 0]
blob_1 = X[y == 1]

print(blob_0.shape)
print(blob_1.shape)

blob_0_y = blob_0[:, 1]
blob_1_y = blob_1[:, 1]

print(blob_0_y.shape)
print(blob_1_y.shape)
```

```python
from sklearn.datasets import make_blobs

X, y = make_blobs(n_samples=500, centers=2, random_state=42)

print(X.shape)
print(y.shape)

blob_0 = X[y == 0]
blob_1 = X[y == 1]

print(blob_0.shape)
print(blob_1.shape)

blob_0_y = blob_0[:, 1]
blob_1_y = blob_1[:, 1]

print(blob_0_y.shape)
print(blob_1_y.shape)

import matplotlib.pyplot as plt

fig, ax = plt.subplots()

ax.boxplot(
    [blob_0_y, blob_1_y],
    whis=(20, 80),
    flierprops=dict(marker='x', markeredgecolor='purple')
)

plt.show()
```

```python
from sklearn.datasets import make_blobs, make_moons, make_circles
import matplotlib.pyplot as plt


# Create datasets
X, y = make_blobs(
    n_samples=100,
    centers=3,
    random_state=42
)

X_moon, y_moon = make_moons(
    n_samples=100,
    random_state=42
)

X_circle, y_circle = make_circles(
    n_samples=100,
    random_state=42
)


# Create 3 subplots
fig, ax = plt.subplots(1, 3)


# Blobs
plot1 = ax[0].scatter(
    X[:, 0],
    X[:, 1],
    marker="s",
    c=y,
    cmap="viridis"
)

# Moons
plot2 = ax[1].scatter(
    X_moon[:, 0],
    X_moon[:, 1],
    marker="o",
    c=y_moon,
    cmap="rainbow"
)

# Circles
plot3 = ax[2].scatter(
    X_circle[:, 0],
    X_circle[:, 1],
    marker="D",
    c=y_circle,
    cmap="coolwarm"
)


# Add colorbars
fig.colorbar(plot1, ax=ax[0])
fig.colorbar(plot2, ax=ax[1])
fig.colorbar(plot3, ax=ax[2])


plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# بارگذاری داده‌های Iris
iris = load_iris()
print(iris.target)

# ترانهاد (Transpose) داده‌ها و ذخیره در متغیر features
features = iris.data.T

# ساخت fig و ax
fig, ax = plt.subplots()

# رسم نمودار scatter
plot = ax.scatter(features[0], features[1], s=features[3] * 200, c=iris.target, cmap="rainbow", alpha=0.5)

# اضافه کردن colorbar
fig.colorbar(plot)

# تنظیم عنوان نمودار
ax.set_title("Iris Dataset")

# نمایش نمودار
plt.show()
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

x = np.array([3, 12, 23, 38, 44, 54, 66, 74, 86, 94]).reshape((-1, 1))

y = np.array([3, 22, 18, 31, 55, 41, 37, 67, 66, 82])

print(x.shape)
print(y.shape)

model = LinearRegression()

model.fit(x, y)
```

```python
y_pred = model.predict(x)

print(y)
print(y_pred)
```

```python
plt.scatter(x, y, color="blue")
```

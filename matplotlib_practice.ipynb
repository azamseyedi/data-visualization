```python
Q1 = df["Phone"].quantile(0.25)
median = df["Phone"].median()
Q3 = df["Phone"].quantile(0.75)

IQR = Q3 - Q1

print(Q1)
print(median)
print(Q3)
print(IQR)

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Phone"] < lower) |
    (df["Phone"] > upper)
]

print(outliers)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Year": [2018, 2019, 2020, 2021, 2022, 2023],
    "India": [1200, 1450, 1100, 1700, 2100, 2500],
    "Iran": [800, 950, 900, 1250, 1600, 1900],
    "Norway": [300, 350, 320, 400, 450, 500]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
df.plot(x="Year", y="India", kind="line")
plt.show()
```

```python
df.plot (x="Year", y="Iran", kind="line")
plt.title("Immigration from Iran")
plt.xlabel("Year")
plt.ylabel("Number of Immigrants")
plt.show()
```

```python
df.plot(x="Year", y=["Iran", "India", "Norway"], kind="line")
plt.title("Immigration from Iran, India and Norway to Canada")
plt.xlabel("Year")
plt.ylabel("Number of Immigrants")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = {
    "Year": [2019, 2020, 2021, 2022, 2023, 2024],
    "Laptop": [120, 150, 180, 210, 260, 310],
    "Phone": [200, 230, 190, 270, 320, 370],
    "Tablet": [80, 95, 110, 105, 140, 160]
}

df = pd.DataFrame(data)

df
```

```python
sns.lineplot(data=df, x="Year", y="Phone")

plt.title("Phone Sales")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.show()
```

```python
sns.lineplot(data=df, x="Year", y="Laptop")

plt.title("Laptop Sales")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.show()
```

```python
sns.lineplot(data=df, x="Year", y="Phone", label="Phone", marker="o")
sns.lineplot(data=df, x="Year", y="Laptop", label="Laptop", marker="o")

plt.title("Phone vs Laptop Sales")
plt.xlabel("Year")
plt.ylabel("Phone vs Laptop")

plt.grid()
plt.legend()
plt.show()
```

```python
sns.lineplot(data=df, x="Year", y="Tablet", label="Tablet", marker="o")

plt.title("Tablet Sales")
plt.xlabel("Year")
plt.ylabel("Sales")

plt.grid()
plt.legend()
plt.show()
```

```python
sns.barplot(data=df, x="Year", y="Tablet")

plt.title("Tablet Sales")
plt.xlabel("Year")
plt.ylabel("Sales")

plt.show()
```

```python
sns.barplot(data=df, x="Year", y="Phone")

plt.title("Phone Sales")
plt.xlabel("Year")
plt.ylabel("Phone")

plt.show()
```
